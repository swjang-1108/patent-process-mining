### merge dataset

```python
event_log = transaction.merge(
    eventcodes,
    how='left',
    left_on='event_code',
    right_on='event_cd'
)
```

```python
event_log
```

```python
event_log = event_log.drop(columns=['event_cd'])
event_log
```

```python
event_log_pm = event_log[
    ['application_number', 'event_code', 'recorded_date', 'event_desc']
].copy()
```

```python
event_log_pm = event_log_pm.sort_values(
    ['application_number', 'recorded_date']
).reset_index(drop=True)
event_log_pm
```

```python
event_log_pm['event_order'] = (
    event_log_pm.groupby('application_number').cumcount() + 1
)
event_log_pm
```

```python
file_path_s = r"C:\Users\ADMIN\DME Dropbox\Park YR\project\Business Process Mining\data\G01_eventlog.tsv"
event_log_pm.to_csv(file_path_s, sep='\t')
```
