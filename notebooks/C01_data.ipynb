```python
# app에서 patent_number 있는 것 = granted
granted = app[app["patent_number"].notna()]["application_number"]
print(f"granted 출원 수: {len(granted):,}")
```

```python
# tx_core를 granted만으로 필터
tx_granted = transaction[transaction["application_number"].isin(granted)]
print(f"tx_granted rows: {len(tx_granted):,}")
```

```python
transaction["recorded_date"] = pd.to_datetime(transaction["recorded_date"], errors='coerce')

nat_apps = transaction[transaction["recorded_date"].isna()]["application_number"].nunique()
total_apps = transaction["application_number"].nunique()

print(f"NaT 있는 application: {nat_apps:,} / {total_apps:,} ({nat_apps/total_apps*100:.2f}%)")

# 어떤 이상한 날짜들이 있는지도 확인
print(transaction[transaction["recorded_date"].isna()]["recorded_date"].head(20))
```

```python
idx_first = transaction.groupby("application_number")["recorded_date"].idxmin()
idx_last = transaction.groupby("application_number")["recorded_date"].idxmax()

first_events = transaction.loc[idx_first, "event_code"]
last_events = transaction.loc[idx_last, "event_code"]

print("=== 첫 번째 event ===")
print(first_events.value_counts().head(20))

print("\n=== 마지막 event ===")
print(last_events.value_counts().head(20))
```
